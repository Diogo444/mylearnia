### Code de récupération du dernier user (fait par ChatGPT)

```python
import mysql.connector

# Connexion à la base MySQL
conn = mysql.connector.connect(
    host="localhost",
    user="Diogo",             # ⚠️ adapte selon ton utilisateur MySQL
    password="Projet_Guichet",         # ⚠️ adapte selon ton mot de passe
    database="projet_guichet"
)

cursor = conn.cursor(dictionary=True)

# Requête SQL : utilisateur le plus récent + infos du guichet
query = """
SELECT u.id_utilisateur, u.nom, u.prenom, u.created_at,
       g.id_guichet, g.libelle_guichet
FROM utilisateurs u
JOIN guichets g ON u.id_guichet = g.id_guichet
ORDER BY u.created_at DESC
LIMIT 1
"""

cursor.execute(query)
dernier_utilisateur = cursor.fetchone()

if dernier_utilisateur:
    print("=== Dernier utilisateur enregistré ===")
    print(f"ID utilisateur : {dernier_utilisateur['id_utilisateur']}")
    print(f"Nom : {dernier_utilisateur['nom']}")
    print(f"Prénom : {dernier_utilisateur['prenom']}")
    print(f"Date création : {dernier_utilisateur['created_at']}")
    print(f"Guichet ID : {dernier_utilisateur['id_guichet']}")
    print(f"Guichet : {dernier_utilisateur['libelle_guichet']}")
else:
    print("Aucun utilisateur trouvé.")

# Fermer la connexion
cursor.close()
conn.close()
```

### Insérer un nouveau user dans la BDD

```python
import mysql.connector
from datetime import datetime

# Connexion à la base MySQL
conn = mysql.connector.connect(
    host="localhost",       # Ton serveur MySQL
    user="Diogo",            # Ton utilisateur (modifie si besoin)
    password="Projet_Guichet",        # Ton mot de passe (modifie si besoin)
    database="projet_guichet"  # Ta base de données
)

cursor = conn.cursor()

def inserer_utilisateur(nom, prenom, id_guichet):
    created_at = datetime.now().strftime('%Y-%m-%d %H:%M:%S')

    sql = "INSERT INTO utilisateurs (nom, prenom, created_at, id_guichet) VALUES (%s, %s, %s, %s)"
    values = (nom, prenom, created_at, id_guichet)

    cursor.execute(sql, values)
    conn.commit() 

    print(f"l'utilisateur {nom} {prenom} a été inséré avec succès")


    cursor.close()
    conn.close()

nom = "Alice"
prenom = "Mortin"
id_guichet = 1

inserer_utilisateur(nom, prenom, id_guichet)
```
