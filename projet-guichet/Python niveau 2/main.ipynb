## Fonction 1 — **calculer_moyenne(notes: list) → float**

```python
notes = []

def calculer_moyenne(notes):
    if len(notes) == 0:
        return 0
    return sum(notes) / len(notes)

while len(notes) < 5:
    try:
        note = float(input("Entrez une note : "))
        notes.append(note)
    except ValueError:
        print("Veuillez entrer un nombre valide.")

print(f"La moyenne est de {round(calculer_moyenne(notes), 2)}")
```

# Cahier des charges : Analyse de notes d'élèves 📝

### 🎯 Objectif général

Créer un programme modulaire en **Python** pour analyser des notes d'élèves. Le programme devra permettre d'ajouter des notes, de calculer des statistiques (moyenne, max, min) et d'afficher un rapport synthétique.

---

### 📌 Contraintes pédagogiques

* Le code doit utiliser au moins **4 fonctions** bien séparées.
* Aucune fonction ne doit contenir de `input()` ou de `print()` sauf celles prévues pour l'affichage.
* Le programme principal doit regrouper les appels à ces fonctions avec des données de test.

---

### 🧩 Fonctionnalités à implémenter

#### `ajouter_note(notes: list, nouvelle_note: float) -> list`

* Ajoute une note à la liste existante et retourne la liste mise à jour.
* **Contrôle :** la note doit être comprise entre **0** et **20**.

#### `calculer_moyenne(notes: list) -> float`

* Calcule et retourne la moyenne des notes, arrondie à **2 décimales**.

#### `trouver_max_min(notes: list) -> tuple`

* Retourne une paire `(max, min)` représentant la note la plus haute et la plus basse.

#### `generer_rapport(notes: list) -> str`

* Retourne une **chaîne de caractères formatée** contenant :
    * La liste des notes.
    * La moyenne.
    * La note maximale.
    * La note minimale.
    * Un message d'appréciation basé sur la moyenne :
        * `si moyenne >= 16` : "Très bien"
        * `si moyenne >= 12` : "Assez bien"
        * `si moyenne >= 10` : "Passable"
        * `sinon` : "Insuffisant"

---

### 🧠 Programme principal

* Initialiser une liste de notes (vide ou pré-remplie).
* Ajouter quelques notes en utilisant la fonction `ajouter_note`.
* Calculer les statistiques (moyenne, max, min).
* Afficher le rapport généré par la fonction `generer_rapport`.

---

### ✅ Notions évaluées

* Définition et appel de fonctions.
* Paramètres, valeur de retour, types `list`, `float`, `tuple`, `str`.
* Gestion d'erreurs simples (note hors limite).
* Manipulation de listes (ajout, analyse).
* Organisation claire du code.


```python
def ajouter_note(notes: list, note: float):
    notes.append(note)

def calculer_moyenne(notes: list):
    return round(sum(notes) / len(notes), 2)


def trouver_max_min(notes: list):
    maximum = max(notes)
    minimum = min(notes)
    return maximum, minimum

def generer_rapport(notes: list):
    moyenne = calculer_moyenne(notes)
    maximum, minimum = trouver_max_min(notes)

    if moyenne >= 16:
        mention = "Très bien"
    elif moyenne >= 12:
        mention = "Assez bien"
    elif moyenne >= 10:
        mention = "Passable"
    else:
        mention = "Insuffisant"

    return (
        f"Liste des notes : {notes}\n"
        f"Note maximale : {maximum}\n"
        f"Note minimale : {minimum}\n"
        f"Moyenne : {moyenne}\n"
        f"Appréciation : {mention}"
    )

def main():
    notes = []

    ajouter_note(notes, 15.5)
    ajouter_note(notes, 18.0)
    ajouter_note(notes, 12.0)
    ajouter_note(notes, 12.5)
    ajouter_note(notes, 5.5)
    ajouter_note(notes, 20)
    ajouter_note(notes, 10.5)
    ajouter_note(notes, 14)
    ajouter_note(notes, 16)
    ajouter_note(notes, 17)
    print(generer_rapport(notes))

main()
```
