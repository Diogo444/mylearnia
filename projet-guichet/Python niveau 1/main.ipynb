|-----------|-----------|

## 👋 Accueil  
Afficher : **« Bienvenue dans l’assistant de profil ! »**

### ✍️ Saisies
- `Prénom` : **str**  
- `Âge` : **int** (vérification numérique)  
- `Taille` : **float** (vérification numérique)  
- `Lieu de naissance` : **str**

### 🧠 Traitements
- Déterminer la variable **majeur**  
  → `True` si âge ≥ 18, `False` sinon  
- Calculer `annee_naissance = 2025 - âge`

### 🗃️ Stockage
- Regrouper les infos dans un **dictionnaire** : `profil`

### 🎨 Affichage personnalisé
- Message : **« Bonjour <prénom> ! »**  
- Année & lieu de naissance  
- Taille + statut majeur/mineur

### 🧩 Message selon l’âge
- `< 12 ans` → « enfant »  
- `12-17 ans` → « ado »  
- `≥ 18 ans` → « adulte »


```python
profil = {}

print("Bienvenue dans l’assistant de profil ! ")

prenom = input("Quel est ton prénom ? : ")
age = int(input("Quel est ton âge ? : "))
taille = float(input("Quelle est ta taille (en centimètre) ? : "))
lieu_de_naissance = input("Quel est ton lieu de naissance ? : ")

est_majeur = age >= 18
annee_naissance = 2025 - age

profil["Prénom"] = prenom
profil["Âge"] = age
profil["Taille"] = taille
profil["Lieu de naissance"] = lieu_de_naissance
profil["Année de naissance"] = annee_naissance
profil["Est majeur"] = est_majeur

print(f"Bonjour {profil['Prénom']} !")
print(f"tu es né le {profil['Année de naissance']} à {profil['Lieu de naissance']}.")

if profil["Est majeur"]:
    print(f"Tu mesures {profil['Taille']} et tu es majeur.")
else:
    print(f"Tu mesures {profil['Taille']}cm et tu es mineur.")

if profil["Âge"] < 12:
    print("Tu es un enfant.")
elif profil["Âge"] >= 12 and profil["Âge"] < 18:
    print("Tu es un adolescent.")
elif profil["Âge"] >= 18:
    print("Tu es un adulte.")
else:
    print("Âge non spécifié.")
```
