# 📚 Listes en Python

### 🔧 Création et lecture

```python
fruits = ["pomme", "banane", "cerise"]
print(fruits)          # ['pomme', 'banane', 'cerise']
print(fruits[0])       # pomme (premier élément)
print(fruits[-1])      # cerise (dernier élément)
```

### ➕ Ajouter un élément

```python
fruits.append("kiwi")
print(fruits)          # ['pomme', 'banane', 'cerise', 'kiwi']
```

### 🗑 Supprimer un élément

```python
fruits.remove("banane")  # supprime par valeur
print(fruits)

dernier = fruits.pop()   # supprime le dernier
print(dernier)           # kiwi
```

### ✏️ Modifier un élément

```python
fruits[0] = "abricot"
print(fruits)          # ['abricot', 'cerise']
```

### 🔁 Parcourir une liste

```python
for fruit in fruits:
    print(fruit.upper())
```

### 📏 Longueur

```python
print(len(fruits))    # nombre d’éléments
```

### 🔍 Rechercher


```python
print("kiwi" in fruits)     # False
```

### 🧪 Exemple complet

```python
notes = [14, 16.5, 12]
moyenne = sum(notes) / len(notes)
print(f"Notes : {notes}")
print(f"Moyenne : {moyenne:.1f}")
```

# 🔒 Tuples en Python
Un tuple est une collection ordonnée mais immuable.
Une fois créé, on ne peut plus le modifier (contrairement aux listes).

### 🔧 Création et lecture

```python
coordonnees = (10, 20)
print(coordonnees)       # (10, 20)
print(coordonnees[0])    # 10
```

✅ Astuce : même avec un seul élément, il faut une virgule → (5,) sinon ce n’est pas un tuple.

### 🛑 Tentative de modification → erreur

```python
coordonnees[0] = 99
# TypeError: 'tuple' object does not support item assignment
```

### 🔁 Parcours d’un tuple

```python
dimensions = (1920, 1080)

for dim in dimensions:
    print(dim)
```

### 📏 Longueur

```python
print(len(dimensions))   # 2
```

### 📦 Déballage (unpacking)

```python
nom, age = ("Alice", 25)
print(nom)     # Alice
print(age)     # 25
```

### ✅ Exemple combiné

```python
film = ("Inception", 2010, "Christopher Nolan")
titre, annee, realisateur = film

print(f"{titre} est sorti en {annee}, réalisé par {realisateur}.")
```

# 📘 Dictionnaires en Python (dict)
Un dictionnaire est une structure qui associe des clés à des valeurs, comme un mini-tableau avec des étiquettes.

### 🔧 Création et lecture

```python
eleve = {
    "nom": "Durand",
    "prenom": "Alice",
    "age": 17
}

print(eleve["nom"])       # Durand
print(eleve.get("prenom"))  # Alice
```

### ✍️ Ajout ou modification

```python
eleve["classe"] = "Terminale"
eleve["age"] = 18
```

### 🗑 Suppression

```python
del eleve["age"]
# ou
eleve.pop("classe")
```

### 🔁 Parcourir un dictionnaire

```python
for cle, valeur in eleve.items():
    print(f"{cle} : {valeur}")
```

### 🔍 Tester la présence d’une clé

```python
if "age" in eleve:
    print("Âge connu :", eleve["age"])
```

### 📏 Longueur

```python
print(len(eleve))    # nombre de paires clé:valeur
```

### ✅ Exemple combiné

```python
produit = {
    "nom": "Clavier",
    "prix": 29.99,
    "stock": 42
}

print(f"{produit['nom']} coûte {produit['prix']} € ({produit['stock']} en stock)")
```

# 🧭 Contrôle de flux : if, elif, else

### 🔧 Structure de base

```python
age = 18

if age < 18:
    print("Mineur")
elif age == 18:
    print("Tout juste majeur")
else:
    print("Majeur confirmé")
```

✅ Indentation obligatoire : 4 espaces ou une tabulation

### 🧠 Conditions possibles

- == : égalité

- != : différent

- <, >, <=, >= : comparateurs

- and, or, not : logique

```python
note = 15

if note >= 10 and note < 16:
    print("Assez bien")
elif note >= 16:
    print("Très bien")
else:
    print("Échec")
```

### 💡 Exemple avec booléens

```python
connecte = True
admin = False

if connecte and admin:
    print("Bienvenue, administrateur.")
elif connecte:
    print("Bienvenue, utilisateur.")
else:
    print("Veuillez vous connecter.")
```

## 🛠 Gestion d’exception : try / except

### 🔧 Structure de base

```python
try:
    # Code qui pourrait planter
    x = int(input("Tape un nombre entier : "))
    print("Tu as saisi :", x)
except:
    print("Ce n'était pas un entier valide.")
```

🧠 Python exécute le bloc try
→ s’il y a une erreur, il saute au except.

### ✅ Version avec type d’erreur précisé


```python
try:
    resultat = 10 / 0
except ZeroDivisionError:
    print("Erreur : division par zéro interdite.")
```

### ✨ Avec message d'erreur système

try:
    num = int("abc")
except ValueError as e:
    print("Erreur :", e)

### 🧹 Optionnel : finally (exécuté quoi qu’il arrive)


```python
try:
    print("Traitement…")
except:
    print("Erreur")
finally:
    print("Nettoyage terminé.")
```
